1. Дослідимо, який філіал супермаркету ('Branch') є найприбутковішим. Для цього знайдіть сумарний прибуток за кожним філіалом і виявіть, який філіал має найвищий.

```python
# Обчислюємо сумарний прибуток для кожного філіалу
total_profit_by_branch = df.groupby('Branch')['Total'].sum()

# Виводимо результат
print(total_profit_by_branch)

# Визначаємо філіал з найвищим прибутком
most_profitable_branch = total_profit_by_branch.idxmax()
print(f'Найприбутковіший філіал: {most_profitable_branch}')
```

2. В якому місті знайходиться філіал з найвищим прибутком? Може в тому місці нам розмітисти ще один магазин.  
Знайдіть відповідь за допомогою функціоналу Pandas.

```python
# Обчислюємо сумарний прибуток для кожного філіалу
total_profit_by_branch = df.groupby('Branch')['Total'].sum()

# Визначаємо філіал з найвищим прибутком
most_profitable_branch = total_profit_by_branch.idxmax()

# Знаходимо місто, де розташований цей філіал
city_of_most_profitable_branch = df[df['Branch'] == most_profitable_branch]['City'].unique()[0]


print(f'Найприбутковіший філіал: {most_profitable_branch}')
print(f'Місто: {city_of_most_profitable_branch}')
```

3.1. Створіть зводну таблицю, яка покаже, скільки покупок (інвойсів) було зроблено в кожній з філій (`Branch`) за різними категоріями товарів. Запишіть таблицю в змінну `invoices_by_category` і виведіть змінну на екран.
Ця таблиця допоможе проаналізувати, в якій філії купують найбільше товарів кожної з категорій.

```python
# Створюємо зведену таблицю
invoices_by_category = df.pivot_table(index='Branch', columns='Product line', values='Invoice ID', aggfunc='count')

display(invoices_by_category)
```

# Гіпотеза 3:
**Чи існує зв'язок між типом клієнта (Customer type) та кількістю придбаних одиниць товару (Quantity)?**


*Питання:*
Чи постійні клієнти (Member) купують більше товарів у порівнянні з новими клієнтами (Normal)?

*Ціль:*
Розробити ефективніші програми лояльності або знижки для залучення нових клієнтів.

```python
# Відповідь на Гіпотезу 1: Оцінка покупця та середній чек

# Обчислюємо середній чек для кожного рівня оцінки (Rating)
average_total_by_rating = df.groupby('Rating')['Total'].mean()


print(average_total_by_rating)
```

Пояснення: Ми групуємо дані за колонкою Rating і обчислюємо середню суму покупки для кожної групи. Це допоможе побачити, як змінюється середній чек залежно від оцінки покупців.

```python
# Відповідь на Гіпотезу 2: Час дня та витрати

# Конвертуємо час в datetime формат
df['Time'] = pd.to_datetime(df['Time'])

# Створюємо нову колонку 'Period', що відображає періоди дня (ранок, день, вечір)
df['Period'] = pd.cut(df['Time'].dt.hour, bins=[0, 12, 18, 24], labels=['Morning', 'Afternoon', 'Evening'], right=False)

# Обчислюємо середню суму покупки для кожного періоду дня
average_total_by_period = df.groupby('Period')['Total'].mean()


print(average_total_by_period)
```

*Пояснення:* Ми розділяємо час на три періоди — ранок, день і вечір — і обчислюємо середній чек у кожному з цих періодів. Це допоможе побачити, в який час дня клієнти витрачають найбільше грошей.

```python
# Відповідь на Гіпотезу 3: Тип клієнта та кількість товарів

# Обчислюємо середню кількість куплених одиниць товару для кожного типу клієнта
average_quantity_by_customer_type = df.groupby('Customer type')['Quantity'].mean()


print(average_quantity_by_customer_type)
```

*Пояснення:* Ми групуємо дані за типом клієнтів (постійні або нові) і обчислюємо середню кількість придбаних товарів. Це покаже, чи постійні клієнти купують більше товарів, ніж нові покупці.
